```python
%sql
SELECT 
ROW_NUMBER()OVER(ORDER BY prd_id) AS product_key,
a.prd_id AS product_id,
a.category_id,
a.product_number,
a.product_name,
a.product_cost,
a.product_line,
b.category,
b.sub_category,
b.maintenance_flag,
a.start_date,
a.end_date
FROM workspace.silver.crm_products AS a
LEFT JOIN workspace.silver.erp_px_cat_g1v2 AS b
ON a.category_id=b.category_id
```
