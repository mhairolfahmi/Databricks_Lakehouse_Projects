```python
%sql
SELECT 
a.order_number,
c.product_key,
b.customer_key,
a.order_date,
a.ship_date,
a.due_date,
a.sales,
a.quantity,
a.price
FROM workspace.silver.crm_sales AS a
LEFT JOIN workspace.gold.dim_customers AS b
ON a.customer_id = b.customer_id
LEFT JOIN workspace.gold.dim_products AS c
ON a.product_number = c.product_number
```

```python
df=spark.sql("""
             SELECT 
                a.order_number,
                c.product_key,
                b.customer_key,
                a.order_date,
                a.ship_date,
                a.due_date,
                a.sales,
                a.quantity,
                a.price
                FROM workspace.silver.crm_sales AS a
                LEFT JOIN workspace.gold.dim_customers AS b
                ON a.customer_id = b.customer_id
                LEFT JOIN workspace.gold.dim_products AS c
                ON a.product_number = c.product_number          
             """)

df.display()
```

```python
df.write.mode("overwrite").format("delta").saveAsTable("workspace.gold.fact_sales")
```
