#Product Performance
* Bike contribute more than 90% of overall sales for the company, while accessories lead in total orders.
* The higher average price of bike contribute to its higher sales compared to others product.
* Australia and United States is the top 2 our the company bikes market shares.
* The sales surge in 2013 was primarily driven by the United States, which recorded the highest sales for most months throughout the year.
* The Mountain-200 is the leading contributor to sales, with all five of the top-selling bikes being variants of this product line.



```python
%sql
--CREATE VIEW FOR ANALYSIS
CREATE OR REPLACE VIEW workspace.gold.monthly_slice AS
SELECT 
DATE_FORMAT(fs.order_date,"yyyy-MM") AS month,
SUM(fs.sales)AS total_sales,
COUNT(DISTINCT fs.order_number) AS total_orders,
dp.category,
dp.product_name,
dc.country
FROM workspace.gold.fact_sales AS fs
INNER JOIN workspace.gold.dim_products AS dp
ON fs.product_key = dp.product_key
INNER JOIN workspace.gold.dim_customers AS dc
ON fs.customer_key = dc.customer_key
GROUP BY month, dp.category,
dp.product_name,
dc.country
ORDER BY month
```

```python
%sql
SELECT *
FROM workspace.gold.monthly_slice
WHERE MONTH IS NOT NULL
```

```python
%sql
SELECT *
FROM workspace.gold.monthly_slice
WHERE  MONTH IS NOT NULL AND category = "Bikes"
```

```python
%sql
SELECT product_name, sum(total_sales) AS total_sales_by_prdt
FROM workspace.gold.monthly_slice
WHERE  MONTH IS NOT NULL AND category = "Bikes"
GROUP BY product_name
ORDER BY total_sales_by_prdt DESC
```
