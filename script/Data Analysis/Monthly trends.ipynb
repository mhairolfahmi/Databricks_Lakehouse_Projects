## Monthly Sales
* monthly sales have grew strongly year over year where the sales performance have accelerated sharply starting from early 2013.
* The number of total orders have growth slowly from 144 in Jan 2011 to 355 in DEC 2012, before it rose sharply to 632 in JAN 2013 and further growth throghout 2013 period.
* while average order value(aov) have show gradual downward trends from year to year.
* This shown that the sales performance of the company is derived from incresing in total orders depsite declining in aov




```python
%sql
-- monthly total sales, total order, 
CREATE OR REPLACE VIEW workspace.gold.monthly_kpi AS
SELECT 
DATE_FORMAT(order_date,"yyyy-MM") AS month,
SUM(sales)AS total_sales,
COUNT(DISTINCT order_number) AS total_orders
FROM workspace.gold.fact_sales
GROUP BY month
```

Monthly Sales

```python
%sql
SELECT *
FROM workspace.gold.monthly_kpi
WHERE MONTH IS NOT NULL
```

Monthly Orders

```python
%sql
SELECT 
month,
total_orders
FROM workspace.gold.monthly_kpi
WHERE month IS NOT NULL
```

Monthly Average Order Value

```python
%sql
SELECT month,
ROUND(total_sales/total_orders,2) AS aov
FROM workspace.gold.monthly_kpi
WHERE month IS NOT NULL
```
